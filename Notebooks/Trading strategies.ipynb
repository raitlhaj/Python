Paying Dividend Strategy

```python

price = 111
discount_rate = 0.08
growth_rate = 0.01

estimated_dividend_yields = [0.08, 0.05, 0.06]

fair_price = estimated_dividend_yields[0]/(1+discount_rate)**1 + estimated_dividend_yields[1]/(1+discount_rate)**2 + (estimated_dividend_yields[2] + price)/(1+discount_rate)**3


estimated_price = 100 * estimated_dividend_yields[0] / (discount_rate - growth_rate)


print('Fair Price:', fair_price,  'with discount rate of', discount_rate*100, '%')
print('Estimated Price:', estimated_price,  'with growth rate of', growth_rate*100, '%')

if( estimated_price > price):
    print('The stock is undervalued')
else:
    print('The stock is overvalued')
```

Index correlation

```python
nasdaq_prices = [100, 102, 101, 105, 107, 110, 108, 111, 115, 117, 120]
nasdaq_returns = [(nasdaq_prices[i] - nasdaq_prices[i-1]) / nasdaq_prices[i-1] for i in range(1, len(nasdaq_prices))]
average_return = sum(nasdaq_returns) / len(nasdaq_returns)

apple_prices = [150, 152, 151, 155, 157, 160, 158, 161, 165, 167, 170]
apple_average_return = sum([(apple_prices[i] - apple_prices[i-1]) / apple_prices[i-1] for i in range(1, len(apple_prices))]) / (len(apple_prices)-1)

estimated_apple_price = apple_prices[9] + apple_average_return * nasdaq_prices[10] 


print('Estimated Apple Price:', estimated_apple_price,  'with average return of', average_return*100, '%')

if( estimated_apple_price > apple_prices[10]):
    print('The stock is undervalued')
else:
    print('The stock is overvalued')
```
