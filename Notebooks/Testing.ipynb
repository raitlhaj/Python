```python
colours1 = ["red", "blue"]
colours2 = colours1
colours3 = colours1.copy() # to create a copy of the list
colours2[1] = "green"
print(colours1)
print(colours2)
print(colours3)
```

```python
from copy import deepcopy
person1 = ["Swen", ["Seestrasse", "Konstanz"]]
person2 = deepcopy(person1)
person2[0] = "Sarah"
person2[1][1] = "Freiburg"
print(person1)
print(person2)
```

```python
dishes = ["pizza", "sauerkraut", "paella", "hamburger"]
countries = ["Italy", "Germany", "Spain", "USA"]
country_specialities_zip = zip(dishes,countries)
print(list(country_specialities_zip))
country_specialities_dict = dict(country_specialities_zip)
country_specialities_dict.items()

country_specialities_dict = dict(zip(["pizza", "sauerkraut", "paella", "hamburger"],["Italy", "Germany", "Spain", "USA"," Switzerland"]))
print(country_specialities_dict)
 
```
