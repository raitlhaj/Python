FORWARD EQUITY

```python
import numpy as np


def Fw_Eqty(S0, r, T, q):
   
    F0 = S0  * np.exp((r-q) * T)
    return F0

Fw_Eqty(100, 0.05, 1, 0.02)
```

CASH DIV

```python
S0 = 100
D = 2 
Snew = (S0 - D)
Snew
```

STOCK DIV

```python
delta = 0.02 # 2 more shares for 100 shares owned
Snew = S0/(1+ delta)
Snew
```

STOCK SPLIT (n for p) p<n

```python
def stock_split(S0, n, p):
    Snew = S0 * (p / n)
    return Snew

stock_split(100, 3, 2)
```
