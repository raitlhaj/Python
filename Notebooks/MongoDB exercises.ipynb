```python
from contextlib import contextmanager
from pymongo import MongoClient
import pandas as pd
import numpy as np
from datetime import datetime
import logging

# Set up logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

# Context manager for MongoDB connection
@contextmanager
def mongo_connection(uri):
    client = MongoClient(uri)
    try:
        yield client
    finally:
        client.close()
        logger.info("MongoDB connection closed")

# Main pipeline function
def financial_data_pipeline(uri, db_name="financial_db", collection_name="tick_data"):
    # Generate sample tick data (like previous sales DataFrame)
    data = pd.DataFrame({
        'timestamp': pd.date_range('2025-01-01', periods=1000, freq='T'),
        'asset': np.random.choice(['stock1', 'stock2', 'stock3'], 1000),
        'price': np.random.uniform(100, 200, 1000),
        'volume': np.random.randint(100, 1000, 1000)
    })
    data['timestamp'] = data['timestamp'].astype(str)  # Convert to string for MongoDB
    
    with mongo_connection(uri) as client:
        db = client[db_name]
        collection = db[collection_name]
        
        # Clear existing data (for tutorial)
        collection.drop()
        
        # Create index for faster queries
        collection.create_index([("asset", 1), ("timestamp", 1)])
        logger.info("Index created on asset and timestamp")
        
        # Insert data
        documents = data.to_dict('records')
        collection.insert_many(documents)
        logger.info(f"Inserted {len(documents)} documents")
        
        # Query: Find high-volume trades (> 800)
        high_volume = list(collection.find({"volume": {"$gt": 800}}).limit(5))
        logger.info(f"High-volume trades: {len(high_volume)} found")
        
        # Aggregation: Mean price per asset
        pipeline = [
            {"$group": {"_id": "$asset", "avg_price": {"$avg": "$price"}}},
            {"$sort": {"_id": 1}}
        ]
        agg_result = list(collection.aggregate(pipeline))
        logger.info(f"Aggregation result: {agg_result}")
        
        # Convert to Pandas for analysis
        agg_df = pd.DataFrame(agg_result).rename(columns={"_id": "asset"})
        agg_df['avg_price'] = agg_df['avg_price'].round(2)
        
        # Query all data to Pandas
        all_data = pd.DataFrame(list(collection.find())).drop(columns=['_id'])
        all_data['timestamp'] = pd.to_datetime(all_data['timestamp'])
        
        return agg_df, all_data

if __name__ == "__main__":
    # Replace with your Atlas URI
    
    try:
        agg_df, full_df = financial_data_pipeline(ATLAS_URI)
        print("Aggregation Results (Mean Price per Asset):\n", agg_df)
        print("\nFirst 5 Rows of Full Data:\n", full_df.head())
    except Exception as e:
        logger.error(f"Error: {e}")
```
