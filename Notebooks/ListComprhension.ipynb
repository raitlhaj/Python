```python
dico ={0:1,1:1}
def fib_ra(n):
    if(n>=0 and type(n)==int):
        if n not in dico:
            dico[n] = fib_ra(n-1) + fib_ra(n-2)
    return dico[n]

ls =[1,4,3,11,5]
list_fib = list(map(fib_ra,ls)) # list comprehension [fib_ra(x) for x in ls]
print(list_fib)

    
```

```python
def rachid_decorator(func):
    def wrapper():
        print("Before the function call")
        func()
        print("After the function call")
    return wrapper


def rachid_decorator2(func):
    def wrapper(*args, **kwargs):
        print("Before the function call")
        res = func(*args, **kwargs)
        print("After the function call")
        return res
    return wrapper

 


@rachid_decorator
def say_hello():
    print("Hello!")
    
    
say_hello()

```
